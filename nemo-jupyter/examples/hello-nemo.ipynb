# Nemo in Jupyter

This notebook uses the **Nemo** kernel: each code cell is a complete Nemo rule program. Running a cell reasons over the program and prints the contents of every `@export`-ed predicate.

Try it: run the next cell, then experiment below.

```nemo
parent(ada, bob) .
parent(bob, cyd) .
parent(cyd, dan) .

ancestor(?x, ?y) :- parent(?x, ?y) .
ancestor(?x, ?y) :- parent(?x, ?z), ancestor(?z, ?y) .

@export ancestor :- csv {}.
```

## Debugging with traces

Kernel commands start with `!` (Nemo reserves `%` for comments). After running a program, `!trace <fact>` shows how a fact was derived:

```nemo
!trace ancestor(ada, dan)
```
